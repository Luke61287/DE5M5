# 1. Load the file

```python
import pandas as pd
import numpy as np

systems_df = pd.read_csv("data/03_Library Systembook.csv")
display(systems_df)
```

# 2. Remove Completely Blank Rows

```python
systems_df = systems_df.dropna(how="all")
display(systems_df)
```

# 3. Remove Extra Quotes from Date Columns

```python
date_columns = ["Book checkout", "Book Returned"]

for col in date_columns:
    systems_df[col] = systems_df[col].astype(str).str.replace('"', '', regex=False)
```

# 4. Trim Whitespace from 'Days allowed to borrow column'

```python
systems_df["Days allowed to borrow"] = systems_df["Days allowed to borrow"].str.strip()
display(systems_df)
```

```python
print(type(systems_df))
```

# 5. Convert Dates to Datetime Format

```python
systems_df["Book checkout"] = pd.to_datetime(systems_df["Book checkout"], dayfirst=True, errors="coerce")

systems_df["Book Returned"] = pd.to_datetime(systems_df["Book Returned"], dayfirst=True, errors="coerce")

print(systems_df)
```

# 6. Identify Invalid Dates

```python
invalid_dates = systems_df[systems_df["Book checkout"].isna()]
print(invalid_dates)
```

# 7. Fix Obvious Year Typos

```python
systems_df["Book checkout"] = systems_df["Book checkout"].apply(lambda x: x.replace(year=2023) if pd.notnull(x) and x.year > 2026 else x)
print(systems_df)
```

# 8. Find Records Returned Before Checkout

```python
bad_dates = systems_df[systems_df["Book Returned"] < systems_df["Book checkout"]]
print(bad_dates)
```

# 9. Remove Duplicate Records

```python
systems_df = systems_df.drop_duplicates(subset = ["Books", "Book checkout", "Book Returned", "Customer ID"])
print(systems_df)
```

# 10. Handle Missing Book Names

```python
systems_df["Books"] = systems_df["Books"].fillna("Unknown")
print(systems_df)
```

# 11. Handle Missing Customer IDs

```python
systems_df["Customer ID"] = systems_df["Customer ID"].fillna(-1)
print(systems_df)
```

# 12. Convert Customer ID to Integer

```python
systems_df["Customer ID"] = (pd.to_numeric(systems_df["Customer ID"], errors="coerce").astype("int64"))
display(systems_df)
```

# 13. Validate Borrowing Period

```python
systems_df["Borrow_Days"] = (systems_df["Book Returned"] - systems_df["Book checkout"]).dt.days
invalid_borrow_period = systems_df[systems_df["Borrow_Days"] > 14]
print(invalid_borrow_period)
```

# 14. Standardise Book Titles


```python
systems_df["Books"] = (systems_df["Books"].str.strip().str.title())
print(systems_df)
```

# 15. Load the Customers file

```python
customers_df = pd.read_csv("data/03_Library SystemCustomers.csv")
print(customers_df)
```

# 16. Remove Completely Empty Rows from Customers file

```python
customers_df = customers_df.dropna(how='all')
print(customers_df)
```

# 17. Remove Records with Missing Customer ID from Customers file

```python
customers_df["DQ_Issues"] = ""

mask = customers_df["Customer ID"].isna()

customers_df.loc[mask, "DQ_Issues"] = (customers_df.loc[mask, "DQ_Issues"] + "MISSING_CUSTOMER_ID;")
display(customers_df)
```

# 5. Trim Whitespace from Customer Names

```python
customers_df["Customer Name"] = customers_df["Customer Name"].str.strip()
display(customers_df["Customer Name"])
```

# 6. Standardise Customer Name Formatting

```python
customers_df["Customer Name"] = customers_df["Customer Name"].str.title()
print(customers_df)
```

# 7. Convert Customer ID to Integer

```python
customers_df["Customer ID"] = (pd.to_numeric(customers_df["Customer ID"], errors="coerce").astype("int64"))
display(customers_df["Customer ID"])
```

# 8. Check for Duplicate Customer IDs

```python
duplicate_customer_ids = customers_df[customers_df.duplicated(subset=["Customer ID"], keep=False)]
print(duplicate_customer_ids)
```

# 9. Write dataframes to Csvs

```python
customers_df.to_csv("data/cleaned_customers.csv", index=False)

systems_df.to_csv("data/cleaned_systems.csv", index=False)
```
